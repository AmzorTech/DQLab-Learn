GMV (Gross Merchandise Value)
- Definisi: Total nilai kotor semua transaksi (penjualan) yang terjadi di platform dalam periode tertentu, sebelum dikurangi potongan, biaya pengembalian, atau komisi.
- Rumus: GMV = Σ (harga jual tiap transaksi × jumlah item)
- Satuan: mata uang (mis. IDR)
- Contoh: Jika platform menjual 3 pesanan masing-masing Rp100.000, Rp250.000, dan Rp150.000, maka GMV = 100.000 + 250.000 + 150.000 = Rp500.000.
- Catatan: GMV mencerminkan volume penjualan kotor, bukan revenue perusahaan setelah biaya. Untuk marketplace, GMV sering dipakai untuk mengukur skala bisnis.

AOV (Average Order Value)
- Definisi: Rata‑rata nilai per pesanan (order) — berapa besar rata‑rata pembelian tiap order.
- Rumus: AOV = GMV / jumlah pesanan
- Satuan: mata uang (mis. IDR)
- Contoh: Dengan GMV Rp500.000 dan 3 pesanan, AOV = 500.000 / 3 ≈ Rp166.667.
- Catatan: AOV membantu mengerti perilaku pembelian; strategi peningkatan AOV (upsell, cross-sell, bundling) bisa menaikkan revenue tanpa menambah jumlah pesanan.

COGS (Cost of Goods Sold)
- Definisi: Biaya langsung yang terkait dengan produksi atau pembelian barang yang dijual — mis. biaya bahan, pembelian dari supplier, biaya produksi per unit.
- Rumus umum: COGS (periode) = awal persediaan + pembelian selama periode – persediaan akhir; atau per item: COGS per unit = biaya per unit.
- Satuan: mata uang (mis. IDR) atau persen (jika dinyatakan terhadap GMV/revenue)
- Contoh: Jika membeli barang seharga Rp50.000 per unit dan menjual 10 unit, total COGS = 50.000 × 10 = Rp500.000.
- Catatan: COGS dipakai untuk menghitung gross profit: Gross Profit = Revenue (atau net sales) – COGS.

Hubungan praktis dan metrik turunannya
- Jika GMV dianggap sebagai total nilai transaksi kotor dan perusahaan mengakui revenue (mis. komisi marketplace atau net sales setelah diskon), maka:
  - Jika revenue = GMV (untuk retailer yang menjual barang sendiri), Gross Margin = (GMV – COGS) / GMV.
  - Jika perusahaan hanya mengambil komisi, revenue perusahaan = persen komisi × GMV.
- Contoh gabungan:
  - GMV = Rp1.000.000 dari 5 order → AOV = 200.000.
  - Jika total COGS untuk barang‑barang tersebut = Rp600.000 → Gross Profit = 1.000.000 – 600.000 = Rp400.000 → Margin = 40%.

Pitfalls & hal yang perlu diperhatikan
- GMV ≠ Profit atau revenue bersih — selalu periksa apakah GMV sudah dikurangi refunds/returns atau belum.
- AOV bisa naik karena kenaikan harga bukan karena lebih banyak item per order — cek juga metric seperti items per order.
- COGS harus menghitung semua biaya langsung; salah hitung COGS bikin margin terlihat lebih baik/buruk dari realita.
- Konsistensi definisi penting ketika membandingkan periode atau kompetitor (mis. apakah GMV termasuk pajak, ongkir, atau dikurangi voucher?).

Contoh perhitungan singkat (Python-like, konsep saja)
- Misal: order_values = [100000, 250000, 150000]
  - GMV = sum(order_values) = 500000
  - AOV = GMV / len(order_values) = 166666.67
  - Jika COGS_total = 300000 → Gross Profit = GMV – COGS_total = 200000

Customer Satisfaction Score (CSAT) adalah metrik yang digunakan untuk mengukur tingkat kepuasan pelanggan terhadap produk atau layanan. Dirumuskan sebagai

  \begin{equation}
  CSAT =  \frac{Total\ Respon\ Positif}{Total\ Respon}*100
  \end{equation}

Dari 125 Responden, 13 diantaranya merespon negatif aplikasi terbaru yang di-deploy oleh DQtech karena ditemukan beberapa *bug*. Dengan bantuan Python, hitung berapa nilai metrik CSAT yang diperoleh oleh DQTech!

````python
# Fill the blank
total_responden = 125
respon_negatif = 13
respon_positif = total_responden - respon_negatif

# Hitung CSAT
# CSAT = (respon_positif / total_responden) * 100
CSAT = respon_positif / total_responden * 100

# Tampilkan CSAT
print(f'Customer Satisfaction Score = {CSAT}%')
print(f'Customer Satisfaction Score = {CSAT:.2f}%')
````
